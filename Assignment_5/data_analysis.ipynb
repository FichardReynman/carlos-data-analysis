```python
# Load pandas and matplotlib, read both Excel files, and check how big each table is.

import pandas as pd
import matplotlib.pyplot as plt

# Initialize variables for channels and videos 
channels = pd.read_excel("youtube_tech_channels_20251120_133753.xlsx") 
videos = pd.read_excel("youtube_tech_videos_20251120_133004.xlsx")

print("Channels:", channels.shape)
print("Videos:", videos.shape)
channels.head() #
```

```python
# Look at the videos table: column names, data types, and the first few rows.

print(videos.columns.tolist())
print()
videos.info()
videos.head()
```

```python
# Check the channels table for missing values and duplicate channel IDs.

channels.info()
print()
print("Missing values:")
print(channels.isna().sum())
print()
print("Duplicate channel IDs:", channels["channel_id"].duplicated().sum())
```

```python
# Convert video dates to real dates and build duration_min (length in minutes) from the raw duration column.

videos["published_at"] = pd.to_datetime(videos["published_at"])
videos["duration_min"] = pd.to_timedelta(videos["duration"]).dt.total_seconds() / 60

print(videos[["duration", "duration_readable", "duration_min"]].head())
print()
print(videos["duration_min"].describe())
```

```python
# Convert channel dates from text to real dates and calculate each channel's age in years.

channels["created_date"] = pd.to_datetime(channels["created_date"], format="ISO8601", utc=True)
channels["scraped_at"] = pd.to_datetime(channels["scraped_at"], format="ISO8601", utc=True)
channels["age_years"] = (channels["scraped_at"] - channels["created_date"]).dt.days / 365.25

print(channels[["channel_name", "created_date", "age_years"]].head())
print()
print(channels["age_years"].describe())
```

```python
# Convert video dates to real dates and build duration_min (length in minutes) from the raw duration column.

videos["published_at"] = pd.to_datetime(videos["published_at"], format="ISO8601", utc=True)
videos["duration_min"] = pd.to_timedelta(videos["duration"]).dt.total_seconds() / 60

print(videos[["duration", "duration_readable", "duration_min"]].head())
print()
print(videos["duration_min"].describe())
```

```python
# Check the videos table for missing values and duplicate video IDs.

videos.info()
print()
print("Missing values:")
print(videos.isna().sum())
print()
print("Duplicate video IDs:", videos["video_id"].duplicated().sum())
```

```python
# Investigate the problem rows: missing video IDs, duplicate IDs, and zero-length videos.

# Rows with missing video_id
print("Missing video_id rows:")
print(videos[videos["video_id"].isna()][["title", "channel_name", "video_url", "views"]])
print()

# Duplicated video_ids (excluding blanks)
dupes = videos[videos["video_id"].notna() & videos["video_id"].duplicated(keep=False)]
print("Duplicated video_id rows:", len(dupes))
print(dupes.sort_values("video_id")[["video_id", "title", "channel_name", "views"]])
print()

# The zero-length videos
print("Zero-length videos:")
print(videos[videos["duration_min"] == 0][["title", "channel_name", "duration", "views"]])
```

```python
# Clean the videos table: recover missing IDs from URLs, drop unaired (zero-length) videos, and fix garbled title text.
# Result is a new table, videos_clean; the original videos table is unchanged.

# 1. Recover missing video IDs from the URL (the ID is whatever follows "v=")
missing = videos["video_id"].isna()
videos.loc[missing, "video_id"] = videos.loc[missing, "video_url"].str.split("v=", n=1).str[1]

# 2. Drop the zero-length (upcoming) videos into a new, cleaned table
videos_clean = videos[videos["duration_min"] > 0].copy()

# 3. Fix the garbled characters in titles (like "Iâ€™ve" -> "I've")
def fix_text(s):
    try:
        return s.encode("cp1252").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return s  # leave it alone if the repair doesn't work

videos_clean["title"] = videos_clean["title"].apply(fix_text)

# Checks
print("Rows:", len(videos), "->", len(videos_clean))
print("Missing IDs left:", videos_clean["video_id"].isna().sum())
print("Duplicate IDs:", videos_clean["video_id"].duplicated().sum())
print()
print(videos_clean.loc[[92, 240], ["video_id", "video_url"]])
print()
print(videos_clean["title"].head())
```

```python
# Group videos into length buckets and compare the number of videos and median views for each.

# How many videos fall in each length bucket?

bins = [0, 1, 5, 10, 15, 25, 60, float("inf")]
labels = ["Under 1 min", "1-5 min", "5-10 min", "10-15 min", "15-25 min", "25-60 min", "Over 60 min"]
videos_clean["length_bucket"] = pd.cut(videos_clean["duration_min"], bins=bins, labels=labels)

print(videos_clean["length_bucket"].value_counts().sort_index())
print()
print(videos_clean.groupby("length_bucket", observed=True)["views"].median())
```

```python
# Adjust for video age and channel size: calculate views per day, compare each video to its channel's typical video, and chart by length.

# 1. How old is each video (in days) on the day of the scrape?
videos_clean["scraped_at"] = pd.to_datetime(videos_clean["scraped_at"], format="ISO8601", utc=True)
videos_clean["age_days"] = (videos_clean["scraped_at"] - videos_clean["published_at"]).dt.total_seconds() / 86400

# 2. Views per day (videos under 1 day old are treated as 1 day to avoid inflated numbers)
videos_clean["views_per_day"] = videos_clean["views"] / videos_clean["age_days"].clip(lower=1)

# 3. Compare each video to its own channel's typical video
channel_median = videos_clean.groupby("channel_id")["views_per_day"].transform("median")
videos_clean["relative_perf"] = videos_clean["views_per_day"] / channel_median.where(channel_median > 0)

# 4. Results by length bucket
results = videos_clean.groupby("length_bucket", observed=True)[["views_per_day", "relative_perf"]].median()
print(results)

# 5. Chart
results["relative_perf"].plot(kind="bar", figsize=(9, 5))
plt.axhline(1, color="gray", linestyle="--")
plt.title("Performance vs. each channel's typical video, by length")
plt.ylabel("Relative performance (1.0 = typical for the channel)")
plt.xlabel("Video length")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Compare topics: tag 5-25 minute videos by title keywords and measure each topic's performance vs. its channel's typical video.

# Focus on videos in a length range comparable to the series
focus = videos_clean[videos_clean["duration_min"].between(5, 25)].copy()
print("Videos in the 5-25 min range:", len(focus))
print()

# Rough keyword groups (you can edit these)
topics = {
    "Python": r"python",
    "JavaScript / web": r"javascript|\bjs\b|react|html|css|web dev|website",
    "AI / ChatGPT": r"\bai\b|chatgpt|gpt|machine learning|llm|neural",
    "Cybersecurity / hacking": r"hack|security|cyber|password|malware",
    "Beginner / tutorial": r"beginner|tutorial|learn|course|how to|explained|crash course",
    "Hardware / gadgets": r"gpu|cpu|pc build|laptop|phone|iphone|nvidia",
    "Games / game dev": r"game|gaming|unity|unreal",
    "Data / SQL": r"data|sql|excel|analytics",
}

rows = []
for name, pattern in topics.items():
    mask = focus["title"].str.contains(pattern, case=False, regex=True, na=False)
    rows.append({
        "topic": name,
        "videos": int(mask.sum()),
        "median_relative_perf": focus.loc[mask, "relative_perf"].median(),
        "median_views_per_day": focus.loc[mask, "views_per_day"].median(),
    })

topic_results = pd.DataFrame(rows).sort_values("median_relative_perf", ascending=False)
print(topic_results.to_string(index=False))

# Chart
topic_results.plot(x="topic", y="median_relative_perf", kind="barh", legend=False, figsize=(9, 5))
plt.axvline(1, color="gray", linestyle="--")
plt.title("Performance vs. channel's typical video, by title topic (5-25 min videos)")
plt.xlabel("Relative performance (1.0 = typical for the channel)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Re-check the topic results with a fairer baseline (each channel's own 5-25 min videos) and add "no topic" and "all videos" 
# comparison rows.

# Same 5-25 min videos, but keep only channels with at least 3 of them
focus = videos_clean[videos_clean["duration_min"].between(5, 25)].copy()
counts = focus.groupby("channel_id")["video_id"].transform("count")
focus = focus[counts >= 3].copy()

# New baseline: each channel's median among ITS OWN 5-25 min videos
base = focus.groupby("channel_id")["views_per_day"].transform("median")
focus["rel_in_range"] = focus["views_per_day"] / base.where(base > 0)

print("Videos used:", len(focus), "from", focus["channel_id"].nunique(), "channels")
print()

rows = []
for name, pattern in topics.items():
    mask = focus["title"].str.contains(pattern, case=False, regex=True, na=False)
    rows.append({"topic": name, "videos": int(mask.sum()),
                 "old_baseline": focus.loc[mask, "relative_perf"].median(),
                 "new_baseline": focus.loc[mask, "rel_in_range"].median()})

any_topic = focus["title"].str.contains("|".join(topics.values()), case=False, regex=True, na=False)
rows.append({"topic": "No topic matched", "videos": int((~any_topic).sum()),
             "old_baseline": focus.loc[~any_topic, "relative_perf"].median(),
             "new_baseline": focus.loc[~any_topic, "rel_in_range"].median()})
rows.append({"topic": "ALL 5-25 min videos", "videos": len(focus),
             "old_baseline": focus["relative_perf"].median(),
             "new_baseline": focus["rel_in_range"].median()})

compare = pd.DataFrame(rows)
print(compare.to_string(index=False))
```

```python
""" Analyze channel growth: calculate subscribers/year, uploads/year, and views per video, 
then test how upload frequency relates to growth. """

ch = channels.copy()
ch["subs_per_year"] = ch["subscribers"] / ch["age_years"].clip(lower=0.5)
ch["videos_per_year"] = ch["total_videos"] / ch["age_years"].clip(lower=0.5)
ch["views_per_video"] = ch["total_views"] / ch["total_videos"].where(ch["total_videos"] > 0)

# Top 10 fastest-growing channels
print(ch.sort_values("subs_per_year", ascending=False)
        [["channel_name", "subscribers", "age_years", "videos_per_year", "views_per_video"]]
        .head(10).round(1).to_string(index=False))
print()

# Do channels that upload more also grow faster? (rank correlation, robust to outliers)
r1 = ch["videos_per_year"].rank().corr(ch["subs_per_year"].rank())
r2 = ch["age_years"].rank().corr(ch["subscribers"].rank())
print("Uploads/year vs subs/year (rank corr):", round(r1, 2))
print("Channel age vs subscribers (rank corr):", round(r2, 2))

# Chart
plt.figure(figsize=(8, 5))
plt.scatter(ch["videos_per_year"], ch["subs_per_year"], alpha=0.6)
plt.xscale("log"); plt.yscale("log")
plt.xlabel("Videos uploaded per year (log scale)")
plt.ylabel("Subscribers gained per year (log scale)")
plt.title("Upload frequency vs. channel growth")
plt.tight_layout()
plt.show()
```

```python
# Combine both datasets: summarize each channel's video lengths, join them to the channel table, and compare growth by format mix.

# Summarize each channel's recent videos
vid_summary = videos_clean.groupby("channel_id").agg(
    median_len_min=("duration_min", "median"),
    share_10_25=("duration_min", lambda s: s.between(10, 25).mean()),
    share_under_3=("duration_min", lambda s: (s < 3).mean()),  # rough proxy for Shorts
).reset_index()

# Attach to the channel table
merged = ch.merge(vid_summary, on="channel_id", how="inner")
print("Channels matched:", len(merged), "of", len(ch))
print()

# Rank correlations with growth
for col in ["median_len_min", "share_10_25", "share_under_3"]:
    r = merged[col].rank().corr(merged["subs_per_year"].rank())
    print(f"{col} vs subs/year (rank corr): {r:.2f}")
print()

# Compare channels by how much of their output is 10-25 min
merged["format_group"] = pd.cut(
    merged["share_10_25"], bins=[-0.01, 0.2, 0.5, 1.0],
    labels=["0-20% of videos", "20-50%", "Over 50%"])
print(merged.groupby("format_group", observed=True).agg(
    channels=("channel_id", "count"),
    median_subs_per_year=("subs_per_year", "median"),
    median_views_per_video=("views_per_video", "median")).round(0))
```

```python
# Chart how much the growth of channels overlaps across the three format groups.

merged.boxplot(column="subs_per_year", by="format_group", figsize=(8, 5))
plt.yscale("log")
plt.suptitle("")
plt.title("Subscribers gained per year, by share of 10-25 min videos")
plt.xlabel("Share of channel's recent videos that are 10-25 min")
plt.ylabel("Subscribers gained per year (log scale)")
plt.tight_layout()
plt.show()
```
