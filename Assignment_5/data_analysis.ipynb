```python
# Import your libraries
import pandas as pd
import matplotlib.pyplot as plt

# Initialize variables for channels and videos 
channels = pd.read_excel("youtube_tech_channels_20251120_133753.xlsx")
videos = pd.read_excel("youtube_tech_videos_20251120_133004.xlsx")

print("Channels:", channels.shape)
print("Videos:", videos.shape)
channels.head()
```

```python
print(videos.columns.tolist())
print()
videos.info()
videos.head()
```

```python
channels.info()
print()
print("Missing values:")
print(channels.isna().sum())
print()
print("Duplicate channel IDs:", channels["channel_id"].duplicated().sum())
```

```python
videos["published_at"] = pd.to_datetime(videos["published_at"])
videos["duration_min"] = pd.to_timedelta(videos["duration"]).dt.total_seconds() / 60

print(videos[["duration", "duration_readable", "duration_min"]].head())
print()
print(videos["duration_min"].describe())
```

```python
channels["created_date"] = pd.to_datetime(channels["created_date"], format="ISO8601", utc=True)
channels["scraped_at"] = pd.to_datetime(channels["scraped_at"], format="ISO8601", utc=True)
channels["age_years"] = (channels["scraped_at"] - channels["created_date"]).dt.days / 365.25

print(channels[["channel_name", "created_date", "age_years"]].head())
print()
print(channels["age_years"].describe())
```

```python
videos["published_at"] = pd.to_datetime(videos["published_at"], format="ISO8601", utc=True)
videos["duration_min"] = pd.to_timedelta(videos["duration"]).dt.total_seconds() / 60

print(videos[["duration", "duration_readable", "duration_min"]].head())
print()
print(videos["duration_min"].describe())
```

```python
videos.info()
print()
print("Missing values:")
print(videos.isna().sum())
print()
print("Duplicate video IDs:", videos["video_id"].duplicated().sum())
```

```python
# Rows with missing video_id
print("Missing video_id rows:")
print(videos[videos["video_id"].isna()][["title", "channel_name", "video_url", "views"]])
print()

# Duplicated video_ids (excluding blanks)
dupes = videos[videos["video_id"].notna() & videos["video_id"].duplicated(keep=False)]
print("Duplicated video_id rows:", len(dupes))
print(dupes.sort_values("video_id")[["video_id", "title", "channel_name", "views"]])
print()

# The zero-length videos
print("Zero-length videos:")
print(videos[videos["duration_min"] == 0][["title", "channel_name", "duration", "views"]])
```
